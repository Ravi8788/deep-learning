```python
# Experiment No. 6
# Aim: Using Transfer Learning in CNN
# Dataset: Caltech-101
# Pretrained Model: ResNet50

# ==========================================================
# 1. IMPORT REQUIRED LIBRARIES
# ==========================================================

import os
import math
import random
import zipfile
import tarfile
import shutil
import urllib.request
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.applications.resnet50 import (
    ResNet50, preprocess_input
)

from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.layers import (
    GlobalAveragePooling2D,
    Dense,
    BatchNormalization,
    Dropout
)

from tensorflow.keras.models import Model

print("TensorFlow Version:", tf.__version__)

# ==========================================================
# 2. DOWNLOAD AND EXTRACT CALTECH-101 DATASET
# ==========================================================

url = "https://data.caltech.edu/records/mzrjq-6wc02/files/caltech-101.zip"
zip_path = "/content/caltech-101.zip"
extract_path = "/content/caltech101"

# Configure custom opener with a browser User-Agent to prevent 403 Forbidden issues
opener = urllib.request.build_opener()
opener.addheaders = [('User-Agent', 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3')]
urllib.request.install_opener(opener)

if not os.path.exists(zip_path):
    print("Downloading Caltech-101 dataset...")
    urllib.request.urlretrieve(url, zip_path)
else:
    print("Dataset zip already downloaded.")

os.makedirs(extract_path, exist_ok=True)

print("Extracting main zip file...")
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

# Find and extract the nested 101_ObjectCategories.tar.gz
nested_tar_path = None
for root, dirs, files in os.walk(extract_path):
    if "__MACOSX" in root:
        continue
    if "101_ObjectCategories.tar.gz" in files:
        nested_tar_path = os.path.join(root, "101_ObjectCategories.tar.gz")
        break

if nested_tar_path is None:
    raise FileNotFoundError("Nested 101_ObjectCategories.tar.gz not found inside the zip.")

print(f"Extracting nested archive: {nested_tar_path}...")
final_extract_dir = "/content/caltech_extracted_images"
os.makedirs(final_extract_dir, exist_ok=True)
with tarfile.open(nested_tar_path, "r:gz") as tar_ref:
    tar_ref.extractall(final_extract_dir)

print("Dataset downloaded and extracted successfully!")

# ==========================================================
# 3. FIND DATASET DIRECTORY
# ==========================================================

data_dir = None

# Search for the directory containing the actual category folders
for root, dirs, files in os.walk(final_extract_dir):
    if "__MACOSX" in root:
        continue
    if "101_ObjectCategories" in dirs:
        data_dir = os.path.join(root, "101_ObjectCategories")
        break
    elif "accordion" in dirs and "airplanes" in dirs:
        data_dir = root
        break

if data_dir is None:
    raise FileNotFoundError(
        "Caltech-101 category folders were not found inside final extraction directory."
    )

print("Dataset directory:", data_dir)

# Remove BACKGROUND_Google if present
background_dir = os.path.join(
    data_dir, "BACKGROUND_Google"
)

if os.path.isdir(background_dir):
    shutil.rmtree(background_dir)

# Find categories containing images
categories = sorted([
    folder for folder in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, folder))
])

print("Number of categories:", len(categories))

# ==========================================================
# 4. SPLIT DATA INTO TRAINING AND TESTING SETS
# ==========================================================

random.seed(42)

test_dir = "/content/caltech_test"

# Create a fresh test directory for repeatable runs
if os.path.exists(test_dir):
    shutil.rmtree(test_dir)

os.makedirs(test_dir, exist_ok=True)

image_extensions = (
    ".jpg", ".jpeg", ".png", ".bmp"
)

for category in categories:

    category_path = os.path.join(data_dir, category)

    test_category_path = os.path.join(
        test_dir, category
    )

    os.makedirs(test_category_path, exist_ok=True)

    images = [
        filename for filename in os.listdir(category_path)
        if filename.lower().endswith(image_extensions)
        and os.path.isfile(
            os.path.join(category_path, filename)
        )
    ]

    random.shuffle(images)

    split = math.floor(len(images) * 0.2)

    test_images = images[:split]

    for image in test_images:
        shutil.move(
            os.path.join(category_path, image),
            os.path.join(test_category_path, image)
        )

print("Training and testing datasets prepared.")

# ==========================================================
# 5. PREPARE IMAGE DATA GENERATORS
# ==========================================================

train_gen = ImageDataGenerator(
    validation_split=0.2,
    preprocessing_function=preprocess_input
)

train_flow = train_gen.flow_from_directory(
    data_dir,
    target_size=(256, 256),
    batch_size=32,
    class_mode="categorical",
    subset="training",
    seed=42
)

valid_flow = train_gen.flow_from_directory(
    data_dir,
    target_size=(256, 256),
    batch_size=32,
    class_mode="categorical",
    subset="validation",
    seed=42
)

test_gen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)

test_flow = test_gen.flow_from_directory(
    test_dir,
    target_size=(256, 256),
    batch_size=32,
    class_mode="categorical",
    classes=list(train_flow.class_indices.keys()),
    shuffle=False
)

num_classes = train_flow.num_classes

print("Number of classes:", num_classes)
print("Training images:", train_flow.samples)
print("Validation images:", valid_flow.samples)
print("Testing images:", test_flow.samples)

# ==========================================================
# 6. LOAD PRETRAINED RESNET50 MODEL
# ==========================================================

res = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(256, 256, 3)
)

# Freeze pretrained convolutional layers
res.trainable = False

print("Pretrained ResNet50 model loaded successfully.")

# ==========================================================
# 7. BUILD THE TRANSFER LEARNING MODEL
# ==========================================================

x = res.output

x = GlobalAveragePooling2D()(x)

x = BatchNormalization()(x)

x = Dropout(0.5)(x)

x = Dense(
    512,
    activation="relu"
)(x)

x = BatchNormalization()(x)

x = Dropout(0.5)(x)

# Output layer for the dataset categories
x = Dense(
    num_classes,
    activation="softmax"
)(x)

model = Model(
    inputs=res.input,
    outputs=x
)

# ==========================================================
# 8. COMPILE THE MODEL
# ==========================================================

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# Display model architecture
model.summary()

# ==========================================================
# 9. TRAIN THE MODEL
# ==========================================================

history = model.fit(
    train_flow,
    epochs=5,
    validation_data=valid_flow
)

# ==========================================================
# 10. EVALUATE THE MODEL
# ==========================================================

test_loss, test_accuracy = model.evaluate(
    test_flow,
    verbose=1
)

print("\nTest Results")
print("Test Loss:", round(test_loss, 4))
print("Test Accuracy:", round(test_accuracy * 100, 2), "%")

# ==========================================================
# 11. PLOT TRAINING AND VALIDATION ACCURACY
# ==========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Transfer Learning using ResNet50")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# ==========================================================
# 12. PLOT TRAINING AND VALIDATION LOSS
# ==========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# ==========================================================
# 13. PREDICT THE CLASS OF A TEST IMAGE
# ==========================================================

test_flow.reset()

images, labels = next(test_flow)

predictions = model.predict(images, verbose=0)

predicted_class = predictions[0].argmax()

class_names = list(test_flow.class_indices.keys())

actual_class = labels[0].argmax()

print("Actual Class:", class_names[actual_class])
print("Predicted Class:", class_names[predicted_class])

plt.figure(figsize=(5, 5))

# Convert ResNet-preprocessed image back to displayable RGB
image = images[0].copy()

image = image[..., ::-1]
image = image + [103.939, 116.779, 123.68]
image = image[..., ::-1]

image = image.clip(0, 255).astype("uint8")

plt.imshow(image)

plt.title(
    "Actual: " + class_names[actual_class]
    + "\nPredicted: " + class_names[predicted_class]
)

plt.axis("off")
plt.show()

# ==========================================================
# 14. SAVE THE TRAINED MODEL
# ==========================================================

model.save("/content/resnet50_transfer_learning.keras")

print("Model saved successfully!")

# ==========================================================
# RESULT
# ==========================================================

print(
    "\nResult: Transfer learning using the pretrained "
    "ResNet50 CNN model was implemented and evaluated "
    "on the Caltech-101 image dataset."
)
```
