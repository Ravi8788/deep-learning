```python
# Experiment No. 10
# Text Generation Using LSTM

# 1. Import required libraries
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

# 2. Create training text
text = """
deep learning is a powerful technique
deep learning is used in artificial intelligence
artificial intelligence is changing the world
machine learning is a part of artificial intelligence
lstm is useful for sequence prediction
lstm can be used for text generation
deep learning models can learn complex patterns
"""

# 3. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total Words:", total_words)

# 4. Create input sequences
input_sequences = []

for line in text.strip().split("\n"):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        sequence = token_list[:i + 1]
        input_sequences.append(sequence)

# 5. Apply padding
max_sequence_len = max(
    len(sequence) for sequence in input_sequences
)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# 6. Separate input and output
X = input_sequences[:, :-1]
labels = input_sequences[:, -1]

y = tf.keras.utils.to_categorical(
    labels,
    num_classes=total_words
)

# 7. Build LSTM model
model = Sequential([
    Embedding(
        input_dim=total_words,
        output_dim=50
    ),
    LSTM(100),
    Dense(total_words, activation="softmax")
])

# 8. Compile model
model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# 9. Display model summary
model.summary()

# 10. Train the model
model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

# 11. Generate new text
seed_text = "deep learning"
next_words = 5

for _ in range(next_words):
    token_list = tokenizer.texts_to_sequences(
        [seed_text]
    )[0]

    token_list = pad_sequences(
        [token_list],
        maxlen=max_sequence_len - 1,
        padding="pre"
    )

    predicted = np.argmax(
        model.predict(token_list, verbose=0),
        axis=-1
    )[0]

    output_word = ""

    for word, index in tokenizer.word_index.items():
        if index == predicted:
            output_word = word
            break

    if not output_word:
        break

    seed_text += " " + output_word

# 12. Display generated text
print("Generated Text:")
print(seed_text)

# 13. Result
print("LSTM text generation completed successfully.")
```
