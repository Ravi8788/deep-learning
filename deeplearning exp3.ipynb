```python
import numpy as np
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

print(X)
```

```python
y = np.array([0, 0, 0, 1])

print(y)
```

```python
weights = np.zeros(2)
bias = 0

print("Weights:", weights)
print("Bias:", bias)
```

```python
learning_rate = 0.1

print("Learning Rate:", learning_rate)
```

```python
def activation(z):
    if z > 0:
        return 1
    else:
        return 0
```

```python
epochs = 20

for epoch in range(epochs):
    total_error = 0

    for i in range(len(X)):

        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Generate prediction
        prediction = activation(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

        total_error += abs(error)

    print("Epoch:", epoch + 1,
          "Weights:", weights,
          "Bias:", round(bias, 2),
          "Error:", total_error)

    # Stop training when there is no error
    if total_error == 0:
        break
```

```python
print("\nFinal Predictions:")

for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    prediction = activation(z)

    print("Input:", X[i],
          "Expected:", y[i],
          "Predicted:", prediction)
```

```python
import numpy as np

# Step 1: Input data for AND gate
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Step 2: Target output
y = np.array([0, 0, 0, 1])

# Step 3: Initialize weights and bias
weights = np.zeros(2)
bias = 0

# Step 4: Learning rate
learning_rate = 0.1

# Step 5: Step activation function
def activation(z):
    if z > 0:
        return 1
    else:
        return 0

# Step 6: Training
epochs = 20

for epoch in range(epochs):
    total_error = 0

    for i in range(len(X)):

        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Generate prediction
        prediction = activation(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

        total_error += abs(error)

    print("Epoch:", epoch + 1,
          "Weights:", weights,
          "Bias:", round(bias, 2),
          "Error:", total_error)

    # Stop training when there is no error
    if total_error == 0:
        break

# Step 7: Testing
print("\nFinal Predictions:")

for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    prediction = activation(z)

    print("Input:", X[i],
          "Expected:", y[i],
          "Predicted:", prediction)
```
